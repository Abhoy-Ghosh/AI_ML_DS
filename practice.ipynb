```python
# Importing dataset & required libraries
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
tips = sns.load_dataset("tips")

# 1. Swarmplot
sns.swarmplot(x='day', y='total_bill', data=tips)
plt.title("Swarm Plot of Total Bill by Day")
plt.show()

# 2. Pairplot
sns.pairplot(tips)
plt.suptitle("Pair Plot of Tips Dataset", y=1.02)
plt.show()

# 3. PairGrid
g = sns.PairGrid(tips)
g.map_upper(sns.scatterplot)
g.map_lower(sns.regplot)
g.map_diag(sns.histplot)
g.fig.suptitle("PairGrid Visualization", y=1.02)
plt.show()

# 4. FacetGrid
g = sns.FacetGrid(tips, col='time')
g.map(sns.scatterplot, 'total_bill', 'tip')
g.fig.suptitle("FacetGrid of Total Bill vs Tip by Time", y=1.05)
plt.show()

# 5. Scatterplot
sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.title("Scatter Plot of Total Bill vs Tip")
plt.show()

# 6. Countplot
sns.countplot(x='day', data=tips)
plt.title("Count of Customers by Day")
plt.show()

# 7. Barplot
sns.barplot(x='day', y='tip', data=tips, estimator=np.mean)
plt.title("Average Tip Amount by Day")
plt.show()

# 8. Violinplot
sns.violinplot(x='day', y='total_bill', data=tips)
plt.title("Distribution of Total Bill by Day (Violin Plot)")
plt.show()

# 9. Heatmap (fixed)
numeric_tips = tips.select_dtypes(include=[np.number])
sns.heatmap(numeric_tips.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap of Numerical Features")
plt.show()
```
