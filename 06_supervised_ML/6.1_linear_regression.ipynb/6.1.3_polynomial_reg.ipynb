```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
# import random

%matplotlib inline
```

```python
np.random.seed(46)

X = 6 * np.random.rand(100,1) - 3
y = 0.5 * X**2 + 1.5 * X + 2 +np.random.randn(100,1)

# eq : y =0.5x^2 + 1.5x + 2 +Outliers

```

```python
plt.scatter(X,y,color ='g')
```

```python
(X,y)
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y, test_size =0.3,random_state=45)
```

```python
# simple linear reg
from sklearn.linear_model import LinearRegression

reg_1= LinearRegression()

reg_1.fit(X_train,y_train)
```

```python
from sklearn.metrics import r2_score

y_pred = reg_1.predict(X_test)
score = r2_score(y_test,y_pred)

print(score)
```

```python
plt.plot(X_train,reg_1.predict(X_train),color ="r")
plt.scatter(X_train,y_train)

plt.xlabel('X dataset')
plt.ylabel('y')
```

```python
# lets apply polynomial transformation

from  sklearn.preprocessing import PolynomialFeatures
```

```python
poly = PolynomialFeatures(degree = 2, include_bias =True)

X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)
```

```python
X_train_poly # first is bias second is x1, third is x^2
```

```python
reg_poly = LinearRegression()
reg_poly.fit(X_train_poly,y_train)
y_pred = reg_poly.predict(X_test_poly)

score = r2_score(y_test,y_pred)
print(score)
```

```python
print(reg_poly.coef_)
```

```python
print(reg_poly.intercept_)
```

```python
plt.scatter(X_train, reg_poly.predict(X_train_poly),color = 'r')
plt.scatter(X_train,y_train)
```

```python
poly = PolynomialFeatures(degree = 3, include_bias =True)

X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)
```

# Pipeline

```python
from sklearn.pipeline import Pipeline
```

```python
def poly_regression(degree):

    poly_features = PolynomialFeatures(degree = degree, include_bias = True)
    lin_reg = LinearRegression()

    poly_regression = Pipeline(
        [
            ('poly_features', poly_features),
            ('linear_regression',lin_reg)
        ]
    )

    poly_regression.fit(X_train, y_train)# polynomial and fit of linear regression

    y_pred_new = poly_regression.predict(X_new)
# plotting prediction line
    plt.plot(X_new,y_pred_new,'r', label = 'Degree' + str(degree),linewidth = 2)
    plt.plot(X_train, y_train, 'b.')
    plt.plot(X_test, y_test, 'g.')

    plt.legend(loc = 'upper left')
    plt.xlabel('x')
    plt.ylabel('y')
```

```python
poly_regression(3)
```
