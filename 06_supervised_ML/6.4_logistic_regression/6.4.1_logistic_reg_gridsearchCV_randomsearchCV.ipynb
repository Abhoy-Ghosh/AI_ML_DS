# Logistic Regression


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 

%matplotlib inline
```

```python
from sklearn.datasets import make_classification

```

```python
# make datasets

X,y=make_classification(n_samples=1000,n_classes=2,n_features=10,random_state=20)
```

```python
pd.DataFrame(X)
```

```python
y # binary classification
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y, test_size=.3,random_state=10)
```

```python
# model training

from  sklearn.linear_model import LogisticRegression

logistic_reg =LogisticRegression()
```

```python
logistic_reg.fit(X_train,y_train)
```

```python
y_pred = logistic_reg.predict(X_test)

y_pred
```

```python
logistic_reg.predict_proba(X_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report

score = accuracy_score(y_test,y_pred)
print(score)

print(confusion_matrix(y_test,y_pred))

print(classification_report(y_test,y_pred))
```

# GridSearch CV

```python
model =  LogisticRegression()
penalty = ['l2','l1','elasticnet']
c_value = [100,10,1.0,0.1,0.01]
solver =  ['newton-cg', 'lbfgs','liblinear', 'sag', 'saga']
```

```python
param =dict(penalty= penalty, C = c_value, solver = solver)
print(param)
```

```python
# # penalty = ['l2', 'l1', 'elasticnet']
# # solver = ['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga']

# # 👉 This will cause invalid combinations like:

# # 'l1' with 'lbfgs' ❌
# # 'elasticnet' without 'saga' ❌


# param_grid = [
#     {
#         'penalty': ['l2'],
#         'solver': ['newton-cg', 'lbfgs', 'sag'],
#         'C': [100, 10, 1.0, 0.1, 0.01]
#     },
#     {
#         'penalty': ['l1'],
#         'solver': ['liblinear', 'saga'],
#         'C': [100, 10, 1.0, 0.1, 0.01]
#     },
#     {
#         'penalty': ['elasticnet'],
#         'solver': ['saga'],
#         'C': [100, 10, 1.0, 0.1, 0.01],
#         'l1_ratio': [0.5]
#     }
# ]
```

```python
# Straitified  K fold
from sklearn.model_selection import StratifiedKFold

cv  = StratifiedKFold(n_splits=5,shuffle=False) # Cross Validation
```

```python
## gridSearchCV

from sklearn.model_selection import GridSearchCV

grid = GridSearchCV(estimator=model,param_grid=param,scoring= 'accuracy',cv =cv, n_jobs =-1)
```

```python
grid
```

```python
X_train.shape
```

```python
print(type(y_train))
print(np.shape(y_train))
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred_grid = grid.predict(X_test) 
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report

score = accuracy_score(y_test,y_pred_grid)
print(score)

print(confusion_matrix(y_test,y_pred_grid))

print(classification_report(y_test,y_pred_grid))
```

# Randomized SearchCV
