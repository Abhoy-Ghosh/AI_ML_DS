# ❌ Error in `make_classification`

## 🚨 Problem

```python
from sklearn.datasets import make_classification

X, y = make_classification(
    n_classes=4,
    n_features=10,
    n_samples=10000,
    random_state=45
)
Error:
ValueError: n_classes(4) * n_clusters_per_class(2) must be smaller or equal 2**n_informative(2)=4
🔍 Why this happens

Default parameters used internally:

n_informative = 2
n_clusters_per_class = 2
Required Constraint
n_classes×n_clusters_per_class≤2
n_informative
Substituting values:
4×2=8
2x2<=4

❌ Since 8 > 4, the configuration is invalid

✅ Fixes
✔ Option 1: Increase n_informative (Recommended)
X, y = make_classification(
    n_classes=4,
    n_features=10,
    n_samples=10000,
    n_informative=3,
    random_state=45
)
✔ Option 2: Reduce n_clusters_per_class
X, y = make_classification(
    n_classes=4,
    n_features=10,
    n_samples=10000,
    n_clusters_per_class=1,
    random_state=45
)
✔ Option 3: Better realistic configuration
X, y = make_classification(
    n_samples=10000,
    n_features=10,
    n_informative=5,
    n_redundant=2,
    n_classes=4,
    n_clusters_per_class=1,
    random_state=45
)
🧠 Key Insight
n_informative controls how many features actually define class separation
Increasing number of classes or clusters requires higher informative dimensions
Otherwise, class separation becomes mathematically impossible
🎯 Rule to Remember

Increase n_informative when increasing n_classes or n_clusters_per_class

````python
X, y = make_classification(
    n_classes=4,
    n_features=10,
    n_samples=10000,
    n_informative=3,
    random_state=45
)
````

````python
pd.DataFrame(X)
````

````python
print(y)
````

````python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y, test_size=.3,random_state=10)
````

````python
from sklearn.linear_model import LogisticRegression

logreg =LogisticRegression()
````

````python
logreg.fit(X_train,y_train)
````

````python
y_pred =  logreg.predict(X_test)
y_pred
````

````python
logreg.predict_proba(X_test)
````

````python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report

score = accuracy_score(y_test,y_pred)
print("accuracy: " ,score)

print(confusion_matrix(y_test,y_pred))

print(classification_report(y_test,y_pred))
````

# GridSearchCV


````python
# GridSearchCV

model_estimator = LogisticRegression()
param = [
    {
        'penalty':['l2'],
        'solver': ['newton-cg','lbfgs','sag'],
        'C':[100, 10, 1.0, 0.1, 0.01]
    },
    {
        'penalty':['l1'],
        'solver': ['liblinear','saga'],
        'C':[100, 10, 1.0, 0.1, 0.01]
    },
    {
        'penalty':['elasticnet'],
        'solver': ['saga'],
        'C':[100, 10, 1.0, 0.1, 0.01]
    }
]

````

````python
from sklearn.model_selection import StratifiedKFold,GridSearchCV,RandomizedSearchCV

gridCV = GridSearchCV(estimator=model_estimator,
                       param_grid=param, 
                       cv = StratifiedKFold(n_splits=5,shuffle=False), 
                       scoring ='accuracy')
````

````python
gridCV.fit(X_train,y_train)
````

````python
gridCV
````

````python
gridCV.best_params_
````

````python
gridCV.best_score_
````

````python
y_pred_grid = gridCV.predict(X_test) 
````

````python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report

score = accuracy_score(y_test,y_pred_grid)
print(score)

print(confusion_matrix(y_test,y_pred_grid))

print(classification_report(y_test,y_pred_grid))
````

# RandomizedSearch CV


````python
# RandomizedSearch CV

randomCV = RandomizedSearchCV(
    estimator= model_estimator,
    param_distributions= param,
    cv= 5,
    scoring= 'accuracy'
)
````

````python
randomCV
````

````python
randomCV.fit(X_train,y_train)
````

````python
y_pred_rand = randomCV.predict(X_test)
````

````python
randomCV.best_params_
````

````python
randomCV.best_score_
````

````python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report

score = accuracy_score(y_test,y_pred_rand)
print(score)

print(confusion_matrix(y_test,y_pred_rand))

print(classification_report(y_test,y_pred_rand))
````

````python
if randomCV.best_score_ > gridCV.best_score_:
    print("yes")
else:
    print("no")
````
